```python
import requests
import time
import os
```

```python
def search_and_download_images(query, num_images, directory):
    try:
        url = 'https://searx.techsaviours.org/'
        params = {
            'q': query,
            'categories': 'images',
            'format': 'json',
            'pageno': 1,
            'count': num_images,
            'language': 'fr'
        }

        image_paths = []

        with requests.session() as session:
            while len(image_paths) < num_images:
                response = session.get(url, params=params)
                response.raise_for_status()  # Check for HTTP request errors

                # Capture cookies
                cookies = response.cookies

                data = response.json()

                if 'results' not in data or len(data['results']) == 0:
                    print(f"No images found for query: {query}")
                    return image_paths

                for index, result in enumerate(data['results']):
                    try:
                        image_url = result['img_src']
                        response = session.get(image_url, cookies=cookies)
                        if response.status_code == 429:
                            print(f"Too many requests. Waiting and retrying...")
                            time.sleep(5)  # Wait for 5 seconds and retry
                            continue  # Retry the request
                        response.raise_for_status()  # Check for HTTP request errors

                        image_path = os.path.join(directory, f'image{len(image_paths)}.jpg')
                        with open(image_path, 'wb') as f:
                            f.write(response.content)
                        image_paths.append(image_path)
                    except requests.exceptions.HTTPError as e:
                        if response.status_code == 404:
                            print(f"Image not found. Skipping...")
                        elif response.status_code == 403:
                            print(f"Forbidden to access the image. Trying next...")
                        else:
                            print(f"An error occurred while downloading image {index}: {e}")
                    except requests.exceptions.RequestException as e:
                        print(f"An error occurred while downloading image {index}: {e}")

        return image_paths

    except requests.exceptions.RequestException as e:
        print(f"An error occurred during image search and download: {e}")
        return []
```

```python
import requests
import os
import time

def search_and_download_images(query, num_images, directory):
    try:
        url = 'https://s.mble.dk/'

        params = {
            'q': query,
            'category_images': '',
            'language': 'fr',
            'safesearch': '0',
            'theme': 'simple'
        }

        image_paths = []

        headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.36'
        }

        with requests.session() as session:
            response = session.get(url + 'search', params=params, headers=headers)
            response.raise_for_status()  # Check for HTTP request errors

            data = response.json()

            if 'results' not in data or len(data['results']) == 0:
                print(f"No images found for query: {query}")
                return image_paths

            for index, result in enumerate(data['results']):
                try:
                    image_url = result['img_src']
                    response = session.get(image_url, headers=headers)
                    response.raise_for_status()  # Check for HTTP request errors

                    image_path = os.path.join(directory, f'image{index}.jpg')
                    with open(image_path, 'wb') as f:
                        f.write(response.content)
                    image_paths.append(image_path)

                    # Introduce a delay between requests to avoid hitting rate limits
                    time.sleep(1)  # Adjust the delay time as needed
                except requests.exceptions.HTTPError as e:
                    print(f"An error occurred while downloading image {index}: {e}")
                except requests.exceptions.RequestException as e:
                    print(f"An error occurred while downloading image {index}: {e}")

        return image_paths

    except requests.exceptions.RequestException as e:
        print(f"An error occurred during image search and download: {e}")
        return []

# Example usage:
search_and_download_images("femme", 3, "images_directory")
```
